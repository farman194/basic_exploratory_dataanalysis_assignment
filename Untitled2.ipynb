##  data loading and printing first 10 rows

```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv(r"C:\Users\Farman Ali\Documents\semester 3\ICA\crop_yield.csv")
print(df.head(10))
```

## checking the missing values in each column

```python
print(df.isnull().sum())
```

## printing datatypes of all columns

```python
print(df.dtypes)
```

## getting statistical summary of all numeric columns

```python
print(df.describe())
```

## getting statistical summary of all objective columns

```python
print(df.describe(include=["object", "string"]))
```

## removing all row where yield_kg column is missing

```python
df = df.dropna(subset=["YIELD_KG"])
```

## filling missing values in Area_Acres columns with average of that column

```python
df["AREA_ACRES"] = df["AREA_ACRES"].fillna(df["AREA_ACRES"].mean())
```

## replacing missing values in district column with string unknown

```python
df["DISTRICT"] = df["DISTRICT"].fillna("Unknown")
```

## replacing missing values in fertilizer_used column with string No

```python
df["FERTILIZER_USED"] = df["FERTILIZER_USED"].fillna("No")
```

## replacing missing values in yield_value column with average of that column

```python
df["YIELD_VALUE"] = df["YIELD_VALUE"].fillna(df["YIELD_VALUE"].mean())
```

## converting harvest_date column in datetime format

```python
df["HARVEST_DATE"] = pd.to_datetime(df["HARVEST_DATE"], errors="coerce")
```

## removing the letter F from farm_ID column and converting the result to integer 

```python
df["FARM_ID"] = (
    df["FARM_ID"].astype(str).str.replace("F", "", regex=False).str.strip().astype(int)
```

## rechecking missing values in data again

```python
print(df.isnull().sum())
```

## displaying datatypes after cleaning

```python
print(df.dtypes)
```

## calculating total yield of each type of crop

```python
crop_col = "CROP_TYPE" if "CROP_TYPE" in df.columns else "CROP"
total_yield_by_crop = df.groupby(crop_col)["YIELD_KG"].sum().sort_values(ascending=False)
print(total_yield_by_crop)
```

## plotting a bar chart of total yield by crop type

```python
plt.figure(figsize=(10, 6))
total_yield_by_crop.plot(kind="bar", color="seagreen", edgecolor="black")
plt.title("Total Yield by Crop Type (KG)")
plt.xlabel("Crop Type")
plt.ylabel("Total Yield (KG)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## creating a pivot table summarizing total yield by harvest date

```python
pivot_yield_by_date = df.pivot_table(
    values="YIELD_KG", index="HARVEST_DATE", aggfunc="sum", fill_value=0
)
print(pivot_yield_by_date)
```
